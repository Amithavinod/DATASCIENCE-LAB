```python
import numpy as np
import matplotlib.pyplot as plt
x=np.array([0,50])
y=np.array([0,50])
plt.subplot(2,3,1)
plt.xlabel("x label",color="blue")
plt.ylabel("y label",color="blue")

plt.title("title 1")
plt.scatter(x,y)

x=np.array([0,50,0,50])
y=np.array([8,16,8,16])
plt.subplot(2,3,2)

plt.plot(x,marker='*',color="orange",linewidth=2.5)
plt.plot(y,color="blue",linestyle="--")

x=np.array([0,15,0,15])
plt.subplot(2,3,3)
plt.plot(x,color="blue",marker='o',mfc="yellow",mec="black")

x=np.array([0,5,10,15,20])
plt.title("title 2")
plt.subplot(2,3,4)

plt.plot(x)
plt.grid()

x=np.array([5,10,3,20,15,30,1,2])
y=np.array([30,35,30,15,10,5,77,33])
plt.subplot(2,3,5)
size=np.array([400,40,40,40,100,100,400,400])
colors=np.array(["red","black","pink","brown","hotpink","green","yellow","violet"])

plt.scatter(x,y,s=size,c=colors,alpha=0.5)

x=np.array(["A","B","C"])
y=np.array([44,55,22])
plt.subplot(2,3,6)
plt.bar(x,y,color="red",width=0.1)

plt.suptitle("GRAPH")
plt.show()
```

```python
y=np.array([35,20,20,25])
mylabels=["35","20","20","25"]
myexplode=[0,0.2,0,0]
plt.pie(y,labels=mylabels,startangle=90,explode=myexplode,shadow=True)
plt.legend()
plt.show()
```
