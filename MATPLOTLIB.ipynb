```python
#matplotlib markers
import matplotlib.pyplot as plt
import numpy as np
xpoints=np.array([0,4,8,12])
plt.plot(xpoints,marker='d')
plt.show()
```

```python
#matplotlib markers
#fmt-marker/line/color (o#marker,--#dotted line,g#green)
#ms-marker size
import matplotlib.pyplot as plt
import numpy as np
xpoints=np.array([0,4,7,99,22])
plt.plot(xpoints,'o--g',ms=20)
plt.show()
```

```python
#edge of the marker
#mec-edge of the marker(red in color-r)
#face of the marker
#mfc-face of the marker(yellow in color-y)
xpoints=np.array([0,44,55])
plt.plot(xpoints,'o:g',ms=10,mec='r',mfc='y')
plt.show()
```

```python
#linestyple can be written as ls
xpoints=np.array([0,55,99,90])
plt.plot(ypoints,ls=':',linewidth='30')
plt.show()
```

```python
x = np.array([80, 85, 90, 95, 100, 105, 110, 115, 120, 125])
y = np.array([240, 250, 260, 270, 280, 290, 300, 310, 320, 330])

font1 = {'family':'serif','color':'blue','size':20}
font2 = {'family':'serif','color':'darkred','size':15}

plt.title("Sports Watch Data", fontdict = font1)
plt.xlabel("Average Pulse", fontdict = font2)
plt.ylabel("Calorie Burnage", fontdict = font2)

plt.plot(x, y)
plt.show()
```
