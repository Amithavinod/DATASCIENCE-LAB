```python
#subplot
import numpy as np
import matplotlib.pyplot as plt
x=np.array([3,5,88,55,22])
y=np.array([4,6,657,7,4])
plt.subplot(2,1,1)

plt.plot(x,y)
plt.title("SALES")
#plot2
x=np.array([0,1,2,3])
y=np.array([10,20,30,40])
plt.subplot(2,1,2)

plt.plot(x,y)
plt.title("INCOME")
plt.suptitle("MY SHOP")
plt.show()
```

```python
#matplotlib scatter
x=np.array([5,4,5,564,6,75,876,87,79,21,12,31])
y=np.array([54,46,55,6,6,77,56,6,9,777,547,4655])
plt.scatter(x,y,color="blue")
x=np.array([2000,4000,600,800,1000])
y=np.array([500,1000,1500,2000,2500])
plt.scatter(x,y,color="hotpink")
plt.show()
```

```python
x=np.array([2000,4000,600,800,1000])
y=np.array([500,1000,1500,2000,2500])
colors=np.array([0,10,15,20,25])
sizes=np.array([20,50,100,200,500])
plt.scatter(x,y,c=colors,cmap='viridis',s=sizes)
plt.colorbar()
plt.show()
```

```python
x=np.array([2000,4000,600,800,1000])
y=np.array([500,1000,1500,2000,2500])
colors=np.array([0,10,15,20,25])
sizes=np.array([20,50,100,200,500])
plt.scatter(x,y,c=colors,cmap='viridis',s=sizes,alpha=0.5)
plt.colorbar()
plt.show()
```

```python
x=np.random.randint(100,size=(100))
y=np.random.randint(100,size=(100))
colors=np.random.randint(100,size=(100))
sizes=10*np.random.randint(100,size=(100))
plt.scatter(x,y,c=colors,s=sizes,alpha=0.5,cmap='nipy_spectral')
plt.colorbar()
plt.show()
```

```python
#BARS
x=np.array(["a","b","c"])
y=np.array([3,8,1,])
plt.bar(x,y)
plt.show()
```

```python
x=np.array(["a","b","c"])
y=np.array([3,8,1,])
plt.barh(x,y)
plt.show()
```

```python
#for bar width and for barh its height
x=np.array(["a","b","c"])
y=np.array([3,8,1,])
plt.subplot(1,2,1)
plt.bar(x,y,color="hotpink",width=0.1)
x=np.array(["a","b","c"])
y=np.array([3,8,1,])
plt.subplot(1,2,2)
plt.barh(x,y,color="hotpink",height=0.1)
plt.show()
```

```python
#histogram
x=np.random.normal(170,10,250)
plt.hist(x)
plt.show()
```

```python
#default start from x axis
y=np.array([35,25,25,15])
mylabels=["APPLES","BANANAS","CHERRIES","DATES"]
plt.pie(y,labels=mylabels)
plt.show()
```

```python
y=np.array([35,25,25,15])
mylabels=["APPLES","BANANAS","CHERRIES","DATES"]
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.show()
```

```python
y=np.array([35,25,25,15])
mylabels=["APPLES","BANANAS","CHERRIES","DATES"]
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,explode=myexplode,shadow=True)
plt.show()
```

```python
y=np.array([35,25,25,15])
mylabels=["APPLES","BANANAS","CHERRIES","DATES"]
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.legend()
plt.show()
```

```python
y=np.array([35,25,25,15])
mylabels=["APPLES","BANANAS","CHERRIES","DATES"]
myexplode=[0.2,0,0,0]
plt.pie(y,labels=mylabels,explode=myexplode)
plt.legend(title="four fruits:")
plt.show()
```
